```python
# 리스트 : 가변 데이터 저장
shopping_list = ['apple','banana','cherry']

print(shopping_list)
print(shopping_list[0])

shopping_list[0] = 'kiwi'
print(shopping_list[0])
print(shopping_list)


```

# 리스트 메서드


```python
fruits = ['apple']
fruits.append('banana')

print(fruits)

#insert
print(fruits)
fruits.insert(0,'kiwi')
print(fruits)

# remove
fruits.remove('apple')
print(fruits)

# fruits.remove('grape') ,없어서 안됨
```

### 인덱스

### 슬라이싱


```python
numbers =[0,1,2,3,4,5]

# 1번부터 4번 미만 까지
numbers[1:4]
# 2칸 간격으로 건너띄고
numbers[::2]
# 역순으로 추출
numbers[::-1]
# 앞에서 3개만 추출
numbers[1:4:]
numbers[:3]

numbers[3:]
```

```python
matrix = [
    [1,2,3],
    [4,5,6]
    
]

print(matrix[1])
print(matrix[1][2])
print(len(matrix), len(matrix[0]))
```
